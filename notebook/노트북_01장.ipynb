## (3) 추론형 AI의 등장

image.png

### https://www.aitimes.com/news/articleView.html?idxno=213616

최근에는 **Reasoning(추론형)** AI 기술이 각광 받고 있습니다.

이것은 인간처럼 "스스로 고민하고 검증하는 논리적 사고력"을 익힌 AI를 말합니다.   

예를 들어 엔비디아(NVIDIA)는 최근 자율주행을 위한 추론형 AI 모델인 **알파마요**(Alpamayo)를 공개하였습니다.
- 기존의 자율주행 AI는 블랙박스 모델처럼 행동의 이유를 설명하지 못했지만,

- 알파마요는 *"300m 앞 낙하물 감지!! 안전거리를 확보하기 위해 지금 차선을 변경합니다"* 처럼 행동의 논리적 이유를 설명해 줍니다.
