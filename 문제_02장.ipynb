# 데이터 불러오기

```python
df = pd.read_csv("car.csv", encoding='cp949')
df
```

## 그래프 1

```python
df1=df[ (df['연도']>=2010) & (df['스타일']=='세단') ]
avg_price=df1.groupby("브랜드")[["가격"]].mean()
print(avg_price)

avg_price.plot(kind="bar")

plt.title("브랜드별 평균 판매가격")
plt.ylabel("가격(USD)")
plt.xlabel("브랜드")
plt.xticks(rotation=45)  # 45도 회전
plt.show()
```

## 그래프 2

```python
import seaborn
plt.figure(figsize=(8, 5))
seaborn.scatterplot(x=df["마력"], y=df["가격"], hue=df["브랜드"])
plt.title("자동차 마력과 판매 가격의 관계")
plt.xlabel("마력(HP)"); plt.ylabel("가격(USD)")
plt.show()
```
