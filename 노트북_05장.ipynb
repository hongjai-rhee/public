```python
df=pd.read_csv("샘플_데이터.csv", encoding='cp949')      # 샘플 데이터 파일
print(df); print()
y=np.array(df['y']).reshape(-1,1)                        # 타깃(10x1)
X=np.array(df[['x1', 'x2']])                             # 입력(10x2)
print("타깃 데이터 형태:", y.shape); print
print("입력 데이터 형태:", X.shape); print
W=np.array([0.2, 0.5]).reshape(-1,1)                     # 가중치 벡터 -> 2x1 행렬

print("가중치 형태:", W.shape); print()
b=np.array(1)                                            # 편향 (1x1)
y_pred=X@W+b                                             # 예측 값(10x1)
print("모델 예측 값(처음 5개):")
print(y_pred[:5]); print()
print(f" MSE 손실= {np.mean((y-y_pred)**2): .3f}")       # 손실 계산
```

#### 텐서의 자동미분과 그래디언트 (1번 업데이트)

```python
import tensorflow as tf
np.set_printoptions(precision=5, suppress=True)  # 소숫점 표시를 3자리까지로 제한

# 텐서 변수 정의
x_data = tf.cast(X, tf.float32)
y_data = tf.cast(y, tf.float32)

# 가중치(w)와 절편(b) 랜덤 초기화
w = tf.Variable(tf.random.normal([2,1]))  # 2x1 행렬
b = tf.Variable(tf.random.normal([1,]))   # 1차원 벡터

print(f"처음 가중치 w:\n{w.numpy()}"); print()
print(f"처음 절편 b: {b.numpy()[0]:.3f}"); print()

# 손실 계산 과정을 테이프로 기록
with tf.GradientTape() as tape:
    y_pred = tf.matmul(x_data, w) + b  # 행렬 연산으로 y_pred 계산
    loss = tf.reduce_mean(tf.square(y_pred - y_data)) # MSE 손실 계산
print(f"처음 손실 b: {loss.numpy():.3f}"); print()

# 그래디언트 계산
d_w, d_b = tape.gradient(loss, [w, b])
print(f"w에 대한 그래디언트:\n{d_w.numpy()}"); print()
print(f"b에 대한 그래디언트:\n{d_b.numpy()}"); print()

# 경사하강법
learning_rate=0.05  # 학습율 지정
w.assign_sub(learning_rate * d_w) # 가중치 업데이트
b.assign_sub(learning_rate * d_b) # 편향 업데이트

print(f"경사하강법을 적용한 새로운 가중치 w:\n{w.numpy()}"); print()
print(f"경사하강법을 적용한 새로운 절편 b: {b.numpy()[0]:.3f}")
```

#### 파라미터 최적화(500 스텝 반복)

[실습 05-13]  

```python
import tensorflow as tf
np.set_printoptions(precision=3, suppress=True)  # 소숫점 표시를 3자리까지로 제한

# 텐서 변수 정의
x_data = tf.cast(X, tf.float32)
y_data = tf.cast(y, tf.float32)

# 가중치(w)와 절편(b) 초기화
w = tf.Variable(tf.random.normal([2,1]))   # 2x1 행렬
b = tf.Variable(tf.random.normal([1, ]))   # 1차원 벡터

learning_rate=0.05  # 학습률
w_history = []; b_history = []; loss_history = []  # 결과 기록 리스트 초기화

# 학습 루프
for i in range(500): # # 파라미터를 500번 업데이트
    with tf.GradientTape() as tape:
        y_pred = tf.matmul(x_data, w) + b  # 행렬 연산으로 y_pred 계산
        loss = tf.reduce_mean(tf.square(y_pred - y_data)) # MSE 손실 계산
    dw, db = tape.gradient(loss, [w, b])
    w.assign_sub(learning_rate * dw)
    b.assign_sub(learning_rate * db)

    w_history.append(w.numpy().flatten())
    b_history.append(b.numpy().flatten())
    loss_history.append(loss.numpy())

print(f"--- 학습 완료 ---")
print(f"최종 가중치 w:\n{w.numpy()}"); print()
print(f"최종 절편 b: {b.numpy()[0]:.3f}"); print()
print(f"최종 손실: {loss.numpy():.3f}")

plt.figure(figsize=(8, 3))
plt.plot(loss_history)
plt.title("스텝별 손실 변화")
plt.xlabel("스텝"); plt.ylabel("평균제곱오차"); plt.grid(True)
plt.show()
```

## 스텝별 가중치와 편향의 변화

```python
w_history_np = np.array(w_history)  # 배열로 변환
b_history_np = np.array(b_history)
plt.figure(figsize=(12, 3))
plt.subplot(1,3,1)
plt.plot(w_history_np[:,0], "b-",  label="w1")
plt.legend()
plt.subplot(1,3,2)
plt.plot(w_history_np[:,1], "r--", label="w2")
plt.legend()
plt.subplot(1,3,3)
plt.plot(b_history_np,  "k-.", label="b")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
