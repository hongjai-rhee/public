# **혈압 데이터를 사용한 벡터 변환 실습**

```python
!pip install koreanize_matplotlib
import koreanize_matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss, accuracy_score
```

```python
df=pd.read_csv("혈압.csv", encoding='cp949')
df
```

```python
# 입력과 타깃 지정
X=df[['나이', '체중', '키', '수면시간']]  # 입력
y=df['혈압']                              # 타깃
X
```

```python
# 파라미터 무작위 추출 후 출력
w1=np.random.normal(size=(4, 2)); print("w1"); print(w1); print()
w2=np.random.normal(size=(2, 1)); print("w2"); print(w2); print()
b1=np.random.normal(1); print("b1"); print(b1); print()
b2=np.random.normal(1); print("b2"); print(b2); print()

# 선형 변환 1
x1 = X @ w1 + b1

# 선형 변환 2
y_pred = x1 @ w2 + b2
print("예측값(y_pred)")
print(y_pred)
print()

# 손실 계산
MSE = np.mean((y_pred-y)**2)
print("평균제곱오차 = ", MSE)
```

### 이 코드를 반복 실행하면, 파라미터가 바뀌고 예측값 및 손실이 바뀝니다.
