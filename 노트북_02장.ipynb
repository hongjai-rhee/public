[실습 02-35]

```python
import pandas as pd; import numpy as np
np.random.seed(42)  # (선택) 재현 가능한 결과를 원하면 시드 설정

week=['일', '월', '화', '수', '목', '금', '토']
weeks=week*5  # week 리스트를 5번 반복 -> 총 35개

sales=np.random.randint(500, 1000, 35)

# 딕셔너리 형태로 만들기
data={'요일': weeks, '판매량': sales}
# 데이터프레임 변환
df=pd.DataFrame(data)
df[:5]  # 처음 5행 출력
```

[실습 02-36]

```python
요일_평균=[]

for i in range(7):
  df1=df[df["요일"] == week[i]];
  z=df1["판매량"].mean()
  요일_평균.append(z)

print(요일_평균); print()
pd.DataFrame(요일_평균, index=week, columns=["평균_판매량"])
```

📌 리스트 컴프리헨션(내포; comprehension) 반복문 사용하기
* 반복문의 압축된 형태
* 다음 1분 영상을 시청해 봅시다.
https://www.youtube.com/watch?v=3wN0eVDhrvE


image.png

```python
요일_평균=[df[df["요일"] == w]["판매량"].mean() for w in week]
pd.DataFrame(요일_평균, index=week, columns=["평균_판매량"])
```

📌while 을 사용하는 반복문 예시

```python
i=1
while i <= 10:   # i가 10 이하일 때 반복
    print(i)
    i += 1       # i에 1을 더함, i=i+1과 동일
```

[실습 02-37]

```python
# 판매량의 전체 평균 계산
m=df["판매량"].mean()
print("판매량의 전체 평균=", m)

for i in range(7):   # 요일 순환 (0-->6)
  avg=요일_평균[i]   # 해당 요일의 평균 판매
  if avg >= m+50:    # 조건
    print(f"{week[i]}요일은 판매량이 매우 높습니다! (평균 {avg:.1f})")
```

[실습 02-38]

```python
for i in range(7):
    avg = 요일_평균[i]

    if avg >= m+50:
        print(f"{week[i]}요일은 판매량이 매우 높습니다! (평균 {avg:.1f})")
    elif avg >= m-50:
        print(f"{week[i]}요일은 판매량이 보통 수준입니다. (평균 {avg:.1f})")
    else:
        print(f"{week[i]}요일은 판매량이 낮습니다. (평균 {avg:.1f})")
```

📌f-스트링 사용하기

```python
max_idx=np.argmax(요일_평균)
max_day=week[max_idx]                   # 해당 요일
max_sales=요일_평균[max_idx]
print(f"평균 판매가 가장 높은 요일은 {max_day}요일이고 판매량은 {max_sales:.1f} 이다.")  # 변수값을 출력
```
